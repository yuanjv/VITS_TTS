```python
import os
import pandas as pd
os.environ["HF_ENDPOINT"] = "https://hf-mirror.com"
from transformers import pipeline


# Load the Whisper model for automatic speech recognition
transcriber = pipeline("automatic-speech-recognition", model="openai/whisper-base")

# Path to the E folder
e_folder = "/root/VITS_TTS/genshin_audio/5297/E"

# List audio files (assuming .wav or similar extensions)
audio_files = [f for f in os.listdir(e_folder) if f.endswith(('.wav', '.mp3', '.flac'))]

# Transcribe each audio file
transcripts = []
for audio_file in audio_files:
    audio_path = os.path.join(e_folder, audio_file)
    result = transcriber(audio_path,return_timestamps=True)
    transcripts.append({"file": audio_file.split('.')[0], "transcript": result["text"]})

# Sort transcripts by the integer in the file name
transcripts.sort(key=lambda x: int(x['file']))

transcripts
```

```python
# Create DataFrame and save to index.csv
df = pd.DataFrame(transcripts)
df.to_csv(os.path.join(e_folder, "index.csv"), index=False)
```
